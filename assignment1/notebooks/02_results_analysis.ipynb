## Results from saved runs

The fixed run IDs below keep every table and figure tied to the archived checkpoints. These cells read existing artifacts; they do not train or evaluate models. 

Note: If you want to train and use other models to compare please change the model in run_ids

```python
import json
import pandas as pd

run_ids = {
    "linear": "20260924-124837-513213-m2-baseline",
    "mlp": "20260924-125349-995443-m2-baseline",
    "cnn": "20260924-125826-798568-m2-baseline",
    "rnn": "20260924-130429-733285-m2-baseline",
    "transformer": "20260924-130954-040349-m2-baseline",
}

rows = []
for model, run_id in run_ids.items():
    run_dir = project_root / "results" / "runs" / model / run_id
    training = json.loads((run_dir / "training.json").read_text(encoding="utf-8"))
    evaluation = json.loads((run_dir / "evaluation.json").read_text(encoding="utf-8"))

    rows.append({
        "model": model,
        "best_epoch": training["best_epoch"],
        "completed_epochs": training["completed_epochs"],
        "test_accuracy_%": 100 * evaluation["accuracy"],
        "macro_f1": evaluation["macro_f1"],
        "parameters": evaluation["trainable_parameters"],
        "training_seconds": training["training_time_seconds"],
        "best_validation_loss": training["best_validation_loss"],
        "inference_ms_per_sample": 1000 * evaluation["inference_time_per_sample_seconds"],
    })

comparison = pd.DataFrame(rows).set_index("model")
precision = {
    "test_accuracy_%": 2,
    "macro_f1": 4,
    "training_seconds": 1,
    "best_validation_loss": 4,
    "inference_ms_per_sample": 4,
}
display(comparison.round(precision))
```

## Error analysis

Confusion matrices use true classes as rows and predictions as columns. The following tables compare the largest confusion and per-class recall for all five models.

```python
from torchvision.datasets import FashionMNIST

class_names = FashionMNIST.classes
rows = []

for model, run_id in run_ids.items():
    run_dir = project_root / "results" / "runs" / model / run_id
    evaluation = json.loads((run_dir / "evaluation.json").read_text(encoding="utf-8"))
    matrix = evaluation["confusion_matrix"]

    count, true_id, predicted_id = max(
        (count, true_id, predicted_id)
        for true_id, row in enumerate(matrix)
        for predicted_id, count in enumerate(row)
        if true_id != predicted_id
    )

    rows.append({
        "model": model,
        "true class": class_names[true_id],
        "predicted as": class_names[predicted_id],
        "samples": count,
    })

display(pd.DataFrame(rows).set_index("model"))
```

```python
recall_rows = []

for model, run_id in run_ids.items():
    run_dir = project_root / "results" / "runs" / model / run_id
    evaluation = json.loads((run_dir / "evaluation.json").read_text(encoding="utf-8"))

    for class_id, counts in enumerate(evaluation["confusion_matrix"]):
        recall_rows.append({
            "class": class_names[class_id],
            "model": model,
            "recall_%": 100 * counts[class_id] / sum(counts),
        })

per_class_recall = (
    pd.DataFrame(recall_rows)
    .pivot(index="class", columns="model", values="recall_%")
    .reindex(index=class_names, columns=run_ids)
)

display(per_class_recall.round(1))
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

fig, axes = plt.subplots(2, 3, figsize=(18, 11))

for ax, (model, run_id) in zip(axes.flat, run_ids.items()):
    path = project_root / "results" / "runs" / model / run_id / "figures" / "confusion_matrix.png"
    ax.imshow(mpimg.imread(path))
    ax.set_title(model.upper())
    ax.axis("off")

axes.flat[-1].axis("off")
plt.tight_layout()
plt.show()
```

### Confusion patterns

Shirt was the lowest-recall class for all five models, ranging from
50.9% (Linear) to 71.8% (RNN). For actual Shirt images, T-shirt/top
was the most frequent incorrect prediction in every model. Pullover
and Coat were also common destinations. CNN correctly identified
717 of 1,000 Shirts, compared with 509 for Linear. This shows a
shared weakness on Shirt classification, although the confusion
matrices alone cannot establish *why* the images were confused.

## Learning curves

Validation loss is plotted across completed epochs. Runs that stopped early therefore have shorter curves; each best checkpoint was selected at its minimum validation loss.

```python
fig, ax = plt.subplots(figsize=(9, 5))

for model, run_id in run_ids.items():
    path = project_root / "results" / "runs" / model / run_id / "training.json"
    training = json.loads(path.read_text(encoding="utf-8"))
    losses = training["history"]["validation_loss"]

    ax.plot(range(1, len(losses) + 1), losses, marker="o", label=model)

ax.set(
    title="Validation loss by epoch",
    xlabel="Epoch",
    ylabel="Validation loss",
)
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

## Model comparison and discussion

CNN achieved the highest test accuracy (90.23%) and Macro-F1 (0.902)
with 53,098 parameters. RNN was close at 89.72% accuracy and had a
lower measured inference time (about 0.0033 versus 0.0070 ms per
sample). Linear was the smallest and fastest model, but its accuracy
was lower at 84.23%. In this configuration, Transformer used the most
parameters (803,338) and had the highest measured inference time
without exceeding CNN's accuracy. MLP reached 88.56% accuracy.

These timings measure batched model forward passes on this machine,
not end-to-end application latency. The results describe these saved
runs; they do not establish that one architecture is always better.

### Shared data and training protocol

All five runs used Fashion-MNIST with seed 42. The official 60,000-image
training set was split into 54,000 training and 6,000 validation images
using a stratified 10% split; the official 10,000-image test set remained
separate. Normalization statistics were calculated from the training
subset only.

The runs used batch size 128, AdamW with learning rate 0.001, and a
maximum of 20 epochs. Each run saved the checkpoint with the lowest
validation loss. Early stopping used patience 5 and minimum improvement
0.0001.

The environment recorded in this notebook after training is Python
3.12.10, PyTorch 2.14.0+cu130, Torchvision 0.29.0+cu130,
scikit-learn 1.9.0, and an NVIDIA GeForce RTX 5070 Ti.

## Limitations and conclusion

This comparison uses one saved run per model with one data split and
seed, so small differences may not be consistent across repeated
runs. Results on 28 × 28 grayscale Fashion-MNIST images may not
generalize to real clothing photographs. Inference timings are
specific to this hardware and batched evaluation.

Under this experimental setup, CNN had the highest measured test
accuracy and Macro-F1. However, model choice also depends on size,
speed, and the types of errors that matter for the application.
